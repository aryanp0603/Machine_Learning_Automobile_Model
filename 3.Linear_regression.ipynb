```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Automobile_data - Automobile_data.csv")
```

```python
df.head()
```

```python
df.tail()
```

```python
df['normalized-losses'].replace("?", np.nan, inplace=True)
```

```python
df.isnull().sum()
```

```python
df["normalized-losses"] = pd.to_numeric(df["normalized-losses"], errors='coerce')
```

```python
m = df["normalized-losses"].mean()
```

```python
print("means of normalized-losses: ", m)
```

```python
df["normalized-losses"]=df["normalized-losses"].fillna(m, inplace=True)
```

```python
df.isnull().sum()
```

```python
"?" in df["horsepower"].values
```

```python
(df["horsepower"]=="?").sum()
```

```python
df[df["horsepower"]=="?"]
```

```python
df["horsepower"] = df["horsepower"].replace("?", np.nan)
```

```python
df["horsepower"] = df["horsepower"].astype("float")
```

```python
df.isnull().sum()
```

```python
n = df["horsepower"].mean()
```

```python
print("Mean of horsepower", n)
```

```python
df["horsepower"] = df["horsepower"].fillna(n,inplace=True)
```

```python
df_num = df.select_dtypes(['int64', 'float64'])
df_cat = df.select_dtypes(object)
```

```python
df_num
```

```python
df_cat
```

```python
from sklearn.preprocessing import LabelEncoder

for col in df_cat:
    le = LabelEncoder()
    df_cat[col] = le.fit_transform(df_cat[col])
```

```python
df_cat.head()
```

```python
df_new = pd.concat([df_num, df_cat], axis= 1)
df_new.head()
```

```python
x = df_new.drop('price', axis=1)
y = df_new['price']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,Y_train,Y_test = train_test_split(x, y, test_size=0.3, random_state=1) 
```

```python
X_train.shape,X_test.shape
```

```python
from sklearn.preprocessing import StandardScaler

ss = StandardScaler()

X_train = ss.fit_transform(X_train)
X_test = ss.transform(X_test)
```

```python
X_train
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr = LinearRegression()
```

```python
lr.fit(X_train, Y_train)
```

```python
y_pred = lr.predict(X_test)
```

```python
y_pred[:10]
```

```python
print("the value of m:", lr.coef_[0])
print("the value of b:", lr.intercept_)
```

```python
comparison = pd.DataFrame({ "Actual": Y_test, "Predicted": y_pred }) 
comparison.head(10)
```

```python
from sklearn.metrics import mean_absolute_error
```

```python
mae = mean_absolute_error(Y_test, y_pred) 
print("Mean Absolute Error:", mae)
```

```python
from sklearn.metrics import mean_squared_error
```

```python
mse = mean_squared_error(Y_test, y_pred)
print("Mean Squared Error:", mse)
```

```python
from sklearn.metrics import r2_score
```

```python
r2 = r2_score(Y_test, y_pred)
print("R² Score:", r2)
```
